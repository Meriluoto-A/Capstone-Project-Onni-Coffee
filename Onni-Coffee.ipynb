```python
import matplotlib.pyplot as plt

#********************************************
# Creating Lists
#********************************************

# Monthly data as lists
months = ["January", "February", "March", "April", "May", "June"]
# Store locations as lists
stores = ["New York", "Paris", "Tokyo"]
# Revenue by month and store as lists
new_york_revenue = [26543, 25320, 32889, 39159, 51700, 54446]
paris_revenue = [27314, 25105, 32835, 39478, 52429, 55083]
tokyo_revenue = [27821, 25720, 33111, 40304, 52599, 56957]
# Total monthly revenue as lists
total_monthly_revenue = [81677, 76145, 98834, 118941, 156727, 166485]
# Total revenue by store as lists
store_revenue = [230057, 232244, 236511]
# Total profit by store as lists
store_profit = [164000, 167067, 169795]

#********************************************
# Printing results
#********************************************

print("This is data for Onni Coffee Company")
print()
print("Months:")
print(months)

print("\nTotal Monthly Revenue:")
print(total_monthly_revenue)

print("\nNew York Monthly Revenue:")
print(new_york_revenue)

print("\nParis Monthly Revenue:")
print(paris_revenue)

print("\nTokyo Monthly Revenue:")
print(tokyo_revenue)

print("\nStores:")
print(stores)

print("\nRevenue by Store:")
print(store_revenue)

print("\nProfit by Store:")
print(store_profit)

#********************************************
# Total revenue
#********************************************

print("\nTotal Sales for January to June:")
print(f"${sum(total_monthly_revenue):,}")

#********************************************
# Bar Charts - Total
#********************************************
plt.figure(figsize=(8, 5))
bars = plt.bar(months, total_monthly_revenue, color="skyblue")
plt.title("Onni Coffee Company - Total Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
# Add value labels on bars
for i, revenue in enumerate(total_monthly_revenue):
    plt.text(i, revenue + 2000, f"{revenue:,}", ha="center")
plt.tight_layout()
plt.show()
print()
#********************************************
# Bar Chart - New York
#********************************************
print("New York Sales")
plt.figure(figsize=(8, 5))
plt.bar(months, new_york_revenue, color="lightcoral")
plt.title("Onni Coffee Company - New York Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
for i, revenue in enumerate(new_york_revenue):
    plt.text(i, revenue + 500, f"{revenue:,}", ha="center")
plt.tight_layout()
plt.show()
print()
#********************************************
# Bar Chart - Paris
#********************************************
print("Paris Sales")
plt.figure(figsize=(8, 5))
plt.bar(months, paris_revenue, color="violet")
plt.title("Onni Coffee Company - Paris Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
for i, revenue in enumerate(paris_revenue):
    plt.text(i, revenue + 500, f"{revenue:,}", ha="center")
plt.tight_layout()
plt.show()
print()
#********************************************
# Bar Chart - Tokyo
#********************************************
print("Tokyo Sales")
plt.figure(figsize=(8, 5))
plt.bar(months, tokyo_revenue, color="lightgreen")
plt.title("Onni Coffee Company - Tokyo Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
for i, revenue in enumerate(tokyo_revenue):
    plt.text(i, revenue + 500, f"{revenue:,}", ha="center")
plt.tight_layout()
plt.show()
```
